```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib as mpl
import seaborn as sns
import scipy

plt.style.use('seaborn-v0_8-ticks')   # good baseline

mpl.rc('text', usetex = True)
params= {'text.latex.preamble' : r"\usepackage{amsmath,txfonts}"}
plt.rcParams.update(params)
```

```python
#2D Ising Plots


df_1d = pd.read_csv("data/1D_Ising_Results.csv")
fig, ax = plt.subplots(1, 3, figsize=(16, 5))

print(df_1d.head())
df_1d = df_1d.apply(pd.to_numeric, errors='coerce')


ax[0].errorbar(df_1d["Temperature"], df_1d["Energy"],        yerr=df_1d["Energy Error"], capsize=3,label = "Monte Carlo")
ax[0].plot(df_1d["Temperature"], df_1d["Analytical Energy"],marker="o", linestyle="none", markersize=4, fillstyle="none", label="Analytic")
ax[1].errorbar(df_1d["Temperature"], df_1d["Magnetization"], yerr=df_1d["Magnetization Error"], capsize=3, label = "Monte Carlo")
ax[1].plot(df_1d["Temperature"], df_1d["Analytical Magnetization"], marker="o", linestyle="none", markersize=4, fillstyle="none", label="Analytic")
ax[2].errorbar(df_1d["Temperature"], df_1d["Heat Capacity"], yerr=df_1d["Heat Capacity Error"], capsize=3, label = "Monte Carlo")
ax[2].plot(df_1d["Temperature"], df_1d["Analytical Heat Capacity"], marker="o", linestyle="none", markersize=4, fillstyle="none", label="Analytic")

ax[0].set_xlabel("Temperature"); ax[0].set_ylabel("Energy per Spin")
ax[1].set_xlabel("Temperature"); ax[1].set_ylabel("Magnetization per Spin")
ax[2].set_xlabel("Temperature"); ax[2].set_ylabel("Specific Heat")

ax[0].legend()
ax[1].legend()
ax[2].legend()

plt.tight_layout()
plt.savefig("plots/1D_Ising_Model.svg",dpi=300, bbox_inches='tight',facecolor='white', transparent=False)
plt.show()
```

```python
scaling_sizes = [5, 10, 20, 60]
colors = ['purple', 'cyan', 'yellowgreen', 'red']
T_c = 2.269

colors = mpl.cm.rainbow(np.linspace(0,1,len(scaling_sizes)))
norm = mpl.colors.Normalize(min(scaling_sizes), max(scaling_sizes))

dfs = {}
for L in scaling_sizes:
    df = pd.read_csv(f"data/2d_L_{L}_Ising_results.csv")
    dfs[L] = df.apply(pd.to_numeric, errors='coerce')

fig, ax = plt.subplots(figsize=(16, 5))

for i,L in enumerate(scaling_sizes):
    df = dfs[L]
    label = f"L={L}"
    ax.errorbar(df["Temperature"], df["Heat Capacity"],
                     yerr=df["Heat Capacity Error"],
                     capsize=3, color=colors[i],label=label)

ax.legend()
ax.set_xlabel("Temperature (T)")
ax.set_ylabel("Specific Heat")
ax.grid(True, alpha=0.3)


plt.tight_layout()
plt.savefig("plots/2d_scaling_results.png", dpi=300)
plt.show()
```
