```python
#2D Ising Plots


df_1d = pd.read_csv("data/1D_Ising_Results.csv")
fig, ax = plt.subplots(1, 3, figsize=(16, 5))

print(df_1d.head())
df_1d = df_1d.apply(pd.to_numeric, errors='coerce')


ax[0].errorbar(df_1d["Temperature"], df_1d["Energy"],yerr=df_1d["Energy Error"], capsize=3,label = "Monte Carlo")
ax[0].plot(df_1d["Temperature"], df_1d["Analytical Energy"],marker="o", linestyle="none", markersize=4, fillstyle="none", label="Analytic")
ax[1].errorbar(df_1d["Temperature"], df_1d["Magnetization"], yerr=df_1d["Magnetization Error"], capsize=3, label = "Monte Carlo")
ax[1].plot(df_1d["Temperature"], df_1d["Analytical Magnetization"], marker="o", linestyle="none", markersize=4, fillstyle="none", label="Analytic")
ax[2].errorbar(df_1d["Temperature"], df_1d["Heat Capacity"], yerr=df_1d["Heat Capacity Error"], capsize=3, label = "Monte Carlo")
ax[2].plot(df_1d["Temperature"], df_1d["Analytical Heat Capacity"], marker="o", linestyle="none", markersize=4, fillstyle="none", label="Analytic")

ax[0].set_xlabel("Temperature"); ax[0].set_ylabel("Energy per Spin")
ax[1].set_xlabel("Temperature"); ax[1].set_ylabel("Magnetization per Spin")
ax[2].set_xlabel("Temperature"); ax[2].set_ylabel("Specific Heat")

ax[0].legend()
ax[1].legend()
ax[2].legend()

plt.tight_layout()
plt.savefig("plots/1D_Ising_Model.svg",dpi=300, bbox_inches='tight',facecolor='white', transparent=False)
plt.show()
```

```python
import scipy.interpolate
import scipy.optimize
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

scaling_sizes = [5, 10, 20, 60]
colors = ['royalblue', 'darkorange', 'forestgreen', 'crimson']
T_c = 2.269

# Load dataframes
dfs = {}
for L in scaling_sizes:
    df = pd.read_csv(f"data/2d_L_{L}_Ising_results.csv")
    for col in df.columns[1:]:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    dfs[L] = df

T_values_2D = dfs[scaling_sizes[0]]["Temperature"].values
binder_cumulants = [dfs[L]["Binder Cumulant"].values for L in scaling_sizes]

# Create interpolated functions and find pairwise crossings
funcs = {}
pairwise = []
for idx, size in enumerate(scaling_sizes):
    funcs[size] = scipy.interpolate.interp1d(T_values_2D, binder_cumulants[idx])
for L_i, L_j in zip(scaling_sizes[:-1], scaling_sizes[1:]):
    diff = lambda T, Li=L_i, Lj=L_j: funcs[Li](T) - funcs[Lj](T)
    root = scipy.optimize.brentq(diff, 2.1, 2.5)
    pairwise.append(root)

T_est = np.mean(pairwise)
print(f"T_est = {T_est:.4f}")

# Colorbar setup
norm = mpl.colors.Normalize(vmin=min(scaling_sizes), vmax=max(scaling_sizes))

fig, ax = plt.subplots(1, 2, figsize=(16, 5))

for idx, size in enumerate(scaling_sizes):
    ax[0].plot(T_values_2D, binder_cumulants[idx], color=colors[idx], label=f"L={size}")
ax[0].axvline(x=T_c, color="black", linestyle="dashed", label="$T_c$")
ax[0].set_xlabel("Temperature(T)")
ax[0].set_ylabel("Binder Cumulant")
ax[0].set_title("Binder Cumulant vs Temperature")
ax[0].set_ylim(0.5, 0.9)
ax[0].set_xlim(1.5, 2.5)
ax[0].legend()
ax[0].minorticks_on()
ax[0].grid(True, alpha=0.7, zorder=0)

for idx, size in enumerate(scaling_sizes):
    ax[1].plot(T_values_2D, binder_cumulants[idx], color=colors[idx], label=f"L={size}")
ax[1].axvline(x=T_c, color="black", linestyle="dashed", label="$T_c$")
ax[1].axvline(x=T_est, color="red", linestyle="dashed", label="T Est.")
ax[1].set_xlim(2.25, 2.27)
ax[1].set_ylim(0.58, 0.63)
ax[1].set_xlabel("Temperature(T)")
ax[1].set_ylabel("Binder Cumulant")
ax[1].set_title("Binder Cumulant vs Temperature")
ax[1].legend()
ax[1].minorticks_on()
ax[1].grid(True, alpha=0.7, zorder=0)


plt.tight_layout()
plt.savefig('plots/Binder_Cumulants.svg', dpi=300, bbox_inches='tight',
            facecolor='white', transparent=False)
plt.show()
```
